```python
def count_frequency(arr):
    freq={}
   
    for i in arr:
        if i in freq:
            freq[i] += 1

        else:
            freq[i] = 1

    return freq        
            
        

cf=count_frequency([1,2,3,3,2,2,4,5])
print(cf)

        
```

```python
def first_Non_repeating_char(s):

    freq = {}

    for i in s:
        if i in freq:
            freq[i] += 1
        else:
            freq[i] = 1

    for ch in s:
        if freq[ch] ==1:
            return ch

    return None
    
nrc1 = first_Non_repeating_char("Ranjeet") 
print(nrc1)
nrc1 = first_Non_repeating_char("ranjeert") 
print(nrc1)
                
        
```

```python
def Anagram(str1,str2):
    n=len(str1)
    c=1
    for i in range(n):
        if str1[i] in str2:
            c+=1
            if c==n:
               return "Anagram"
            else:
               return "Not Anagram"
            

a=Anagram('listen','silent')    
print(a)
```

```python
class Dict:
    def __init__(self, size):
        self.size = size
        self.slots = [None]*self.size
        self.data = [None]*self.size

    def put(self,key,value):
        hash_value = self.hash_function(key)
        if self.slots[hash_value] == None:
            self.slots[hash_value] = key
            self.data[hash_value] = value
        else:
            if self.slots[hash_value] == key:
                self.data[hash_value] = value

    def hash_function(self,key):
        return  abs(hash(key)) % self.size
            
```

```python
d=Dict(3)
print(d.slots)
print(d.data)
```
