**Types of BST**



**1. Normal / Basic BST**

A tree where:

-> Left subtree → values smaller than root

-> Right subtree → values greater than root

**2. Balanced BST**

The height of the left and right subtrees is kept relatively close.

```
       50
      /  \
    30    70
   / \    / \
 20  40  60  80


```

🙃Search/insert/delete can be O(log n).

**3. Degenerate / Skewed BST**

When every node has only one child, the BST behaves like a linked list.

-> Time complexity can become O(n).

There are two forms:

• Left-skewed → nodes mainly have left children

• Right-skewed → nodes mainly have right children


image.png

**4. AVL Tree**

A self-balancing BST.

For every node:

Balance Factor = height(left) − height(right)

It must be -1, 0, or +1.



[Further AVL Tree](https://www.geeksforgeeks.org/dsa/introduction-to-avl-tree/)








image.png

**4. Red-Black Tree**

A red-black tree is a kind of **self-balancing binary search tree** where each node has an extra bit, and that bit is often interpreted as the color (red or black). These colors are used to ensure that the tree remains balanced during insertions and deletions. Although the balance of the tree is not perfect, it is good enough to reduce the searching time and maintain it around **O(log n)** time, where n is the total number of elements in the tree.

[Further Red-Black Tree](https://www.geeksforgeeks.org/dsa/introduction-to-red-black-tree/)

```
| Type           | Balanced?       | Search                       |
| -------------- | --------------- | ---------------------------- |
| Normal BST     | Not necessarily | O(log n) average, O(n) worst |
| AVL Tree       | Yes             | O(log n)                     |
| Red-Black Tree | Yes             | O(log n)                     |
| Skewed BST     | No              | O(n)                         |

```
