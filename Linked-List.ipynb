
    # Linked List

## 1. What is a Linked List?

A **Linked List** is a linear data structure where elements are stored in separate objects called **nodes**.

Each node contains:

1. **Data** → stores the value
2. **Next** → stores the reference/address of the next node

Example:

```text
10 → 20 → 30 → 40 → None
```

Here:

* `10` is the first node
* `40` is the last node
* `None` means there is no next node

---

# 2. Structure of a Node

A basic node contains:

```text
┌──────────┬──────────┐
│   Data   │   Next   │
└──────────┴──────────┘
```

In Python:

```python
class Node:

    def __init__(self, data):
        self.data = data
        self.next = None
```

Example:

```python
node1 = Node(10)
node2 = Node(20)

node1.next = node2
```

Now:

```text
10 → 20 → None
```

# 5. Traversing a Linked List

Traversal means visiting every node one by one.

Example:

```text
head
 ↓
10 → 20 → 30 → 40 → None
```

### Code

```python
def traverse(self):

    current = self.head

    while current is not None:
        print(current.data)
        current = current.next
```

### How it works

Initially:

```text
current = head
```

Then:

```text
current.data
```

gets the current node's data.

And:

```python
current = current.next
```

moves to the next node.

# 7. Insert at Beginning

To insert a node at the beginning:


```python
def insert_beginning(self, data):

    new_node = Node(data)

    new_node.next = self.head

    self.head = new_node
```

### Important

```python
new_node.next = self.head
```

Then:

```python

```

# 8. Insert at End

Example:

```text
Before:

10 → 20 → 30 → None


Insert 40:

10 → 20 → 30 → 40 → None
```

### Code

```python
def insert_end(self, data):

    new_node = Node(data)

    if self.head is None:
        self.head = new_node
        return

    current = self.head

    while current.next is not None:
        current = current.next

    current.next = new_node
```

### Important

We move until:

```python
current.next is None
```

This means `current` is the **last node**.

Then:

```python
current.next = new_node
```

connects the new node.

# 9. Insert After a Node

```python
def insert_after(self, key, data):

    current = self.head

    while current is not None:

        if current.data == key:

            new_node = Node(data)

            new_node.next = current.next
            current.next = new_node

            return

        current = current.next
```

### Important

```python
current.next = new_node
```

# 11. Delete from End

Before:

```text
10 → 20 → 30 → None
```

Delete `30`:

```text
10 → 20 → None
```

### Code

```python
def delete_end(self):

    if self.head is None:
        return

    if self.head.next is None:
        self.head = None
        return

    current = self.head

    while current.next.next is not None:
        current = current.next

    current.next = None
```

### Why `current.next.next`?

We need to stop at the **second-last node**.

Example:

```text
10 → 20 → 30 → None
     ↑
   current
```

When:

```python
current.next.next is None
```

`current` is at `20`.

Then:

```python
current.next = None
```

removes `30`.

```python
    current = self.head
```

```python
        if current.data == key:

        current = current.next
```

# 15. Find Middle Node

Example:

```text
10 → 20 → 30 → 40 → 50
```

Middle:

```text
30
```

A common method is **Slow and Fast Pointer**.

```python
def find_middle(self):

    slow = self.head
    fast = self.head

    while fast is not None and fast.next is not None:

        slow = slow.next
        fast = fast.next.next

    return slow
```

### Idea

```text
slow → moves 1 step
fast → moves 2 steps
```

When `fast` reaches the end, `slow` is at the middle.

# 16. Reverse a Linked List

Original:

```text
10 → 20 → 30 → 40 → None
```

After reverse:

```text
40 → 30 → 20 → 10 → None
```

### Code

```python
def reverse(self):

    previous = None
    current = self.head

    while current is not None:

        next_node = current.next

        current.next = previous

        previous = current
        current = next_node

    self.head = previous
```

### The three important variables

```text
previous
current
next_node
```

Initially:

```text
previous = None
current = head
```

Then repeatedly:

```python
next_node = current.next
current.next = previous
previous = current
current = next_node
```

Finally:

```python
self.head = previous
```

# 18. Types of Linked Lists

## Singly Linked List

Each node points only to the next node.

```text
10 → 20 → 30 → None
```

Structure:

```text
[data | next]
```

---

## Doubly Linked List

Each node has:

* Previous pointer
* Data
* Next pointer

```text
None ← 10 ⇄ 20 ⇄ 30 → None
```

Structure:

```text
[prev | data | next]
```

---

## Circular Linked List

The last node points back to the first node.

```text
10 → 20 → 30
↑         ↓
└─────────┘
```

There is no `None` at the end.

---

# 19. Singly vs Doubly Linked List

| Feature          | Singly | Doubly |
| ---------------- | ------ | ------ |
| Next pointer     | Yes    | Yes    |
| Previous pointer | No     | Yes    |
| Move forward     | Yes    | Yes    |
| Move backward    | No     | Yes    |
| Memory           | Less   | More   |

---

# 20. Array vs Linked List

| Feature              | Array              | Linked List    |
| -------------------- | ------------------ | -------------- |
| Memory               | Usually contiguous | Non-contiguous |
| Access by index      | Fast `O(1)`        | Slow `O(n)`    |
| Insert beginning     | Costly `O(n)`      | `O(1)`         |
| Delete beginning     | Costly `O(n)`      | `O(1)`         |
| Search               | `O(n)`             | `O(n)`         |
| Extra pointer memory | No                 | Yes            |

---

# 21. Time Complexity

For a basic singly linked list:

| Operation           |   Time |
| ------------------- | -----: |
| Access by position  | `O(n)` |
| Search              | `O(n)` |
| Insert at beginning | `O(1)` |
| Insert at end       | `O(n)` |
| Delete beginning    | `O(1)` |
| Delete end          | `O(n)` |
| Reverse             | `O(n)` |
| Find middle         | `O(n)` |

If we maintain a **tail pointer**, insertion at the end can be `O(1)`.

````python
class LinkedList:
    def __init__(self):
        
        #create Empty list
        self.head =None
        # no of nodes in LL
        self.n=0

    def __len__(self):
        return self.n

    def insert_head(self,value):

        #new node
        new_node = Node(value)

        #create connection
        new_node.next = self.head

        #reassign head
        self.head = new_node

        #increment n
        self.n = self.n+1

    def __str__(self):    # traverse

         curr = self.head

         result = ''
        
         while curr != None:
             result = result + str(curr.data) + '-->'
             curr = curr.next #updation
             
         return result[:-3]           

    def append(self,value):
        
        new_node = Node(value)

        if self.head==None:
            self.head=new_node
            self.n =self.n+1
            return
        
        curr = self.head
        while curr.next != None :
            curr = curr.next
        # you are at last node    
        curr.next = new_node
        self.n =self.n+1

    def insert_after(self,after, value):

        new_node = Node(value)

        curr = self.head

        while curr != None :
            if curr.data == after:
                break
            curr = curr.next 

            # case 1 break -> item mil gaya -> curr -> not none
        if curr != None:
                #logic
            new_node.next = curr.next
            curr.next = new_node 
            self.n = self.n+1
        else:
            return 'Item not found'
            #case 2 break -> Loop pura chala -> item nii mila -> None


    def clear(self):
         self.head = None
         self.n = 0
         return "LL become Empty"

    def delete_head(self):
        
        if self.head == None :  #empty
            return "Empty LL"
    
        self.head = self.head.next
        self.n = self.n - 1

    def pop(self):

        if self.head == None:
            return "Empty ll"

        curr = self.head

        #kya ll m ek hi item h
        if curr.next == None:
            return self.delete_head()
        
        while curr.next.next != None:  #last ke ek phle stop krna h
            curr = curr.next

        # now curr is at 2nd last position
        curr.next = None
        self.n = self.n - 1

    def remove(self,value):

        if self.head == None:
            return "Empty ll"

        if self.head.data == value:
            # remove to head node
            return self.delete_head()

        curr = self.head
        

        while curr.next != None:

            if curr.next.data == value:
                break
            curr = curr.next

        #item nii mila
        if curr.next == None:
            return "Not Found"
        else:  
            # item mil gaya
            curr.next = curr.next.next    


    def search(self,item):
        
        curr = self.head
        pos = 0

        while curr != None:
            if curr.data == item:
                return pos
            curr = curr.next
            pos = pos + 1
            
        return "Not Found" 

    def __getitem__(self,index):
        
        curr = self.head
        pos=0

        while curr != None:
            if pos ==index:
                return curr.data
            curr = curr.next
            pos = pos + 1
            
        return "Out of Index"    
            
        
           
````
