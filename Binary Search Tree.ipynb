# Binary Search Tree (BST)

## 1. What is a Binary Search Tree?

A **Binary Search Tree (BST)** is a binary tree where:

**Left Subtree < Root < Right Subtree**

Example:

```text
        50
       /  \
     30    70
    /  \   /  \
   20  40 60  80
```

For every node:

* Smaller values → **Left**
* Greater values → **Right**

---

## 2. BST Node

A node contains:

* `data`
* `left`
* `right`

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None
```

Structure:

```text
       Node
      /    \
   left   right
```

---

# 3. Tree Traversals

Traversal means **visiting every node of the tree**.

Consider:

```text
        50
       /  \
     30    70
    /  \   /  \
   20  40 60  80
```

### Inorder

**Left → Root → Right**

```text
20 → 30 → 40 → 50 → 60 → 70 → 80
```

⭐ In a BST, **Inorder gives sorted order**.

### Preorder

**Root → Left → Right**

```text
50 → 30 → 20 → 40 → 70 → 60 → 80
```

### Postorder

**Left → Right → Root**

```text
20 → 40 → 30 → 60 → 80 → 70 → 50
```

### Level Order

Visit nodes **level by level**.

```text
50 → 30 → 70 → 20 → 40 → 60 → 80
```

---

# 4. Insertion

Insertion means **adding a new node to the BST**.

### Rules

```text
New value < Current node → Go LEFT

New value > Current node → Go RIGHT
```

Example: Insert `65`

```text
        50
          \
           70
          /
        60
          \
           65
```

Path:

```text
65 > 50 → Right
65 < 70 → Left
65 > 60 → Right
```

So `65` becomes the right child of `60`.

### Code

```python
def insert(self, root, val):

    if root is None:
        return Node(val)

    if val < root.data:
        root.left = self.insert(root.left, val)

    else:
        root.right = self.insert(root.right, val)

    return root
```

### Important

```python
root.left = self.insert(root.left, val)
```

means:

> Insert into the left subtree and reconnect the updated subtree.

Similarly:

```python
root.right = self.insert(root.right, val)
```

means:

> Insert into the right subtree and reconnect the updated subtree.

# 5. Searching

Searching means **finding whether a value exists in the BST**.

### Rules

```text
value == root.data → Found

value < root.data → Search LEFT

value > root.data → Search RIGHT

root is None → Not Found
```

Example: Search `60`

```text
        50
          \
           70
          /
        60
```

Path:

```text
60 > 50 → Right
60 < 70 → Left
60 == 60 → Found
```

### Code

```python
def search(self, root, key):

    if root is None:
        return False

    if root.data == key:
        return True

    if key < root.data:
        return self.search(root.left, key)

    return self.search(root.right, key)
```

# 6. Minimum Value

In a BST:

**Minimum = Leftmost Node**

Example:

```text
        50
       /
     30
     /
   20
```

Minimum = `20`

### Rule

```text
Go LEFT → LEFT → LEFT → ...
```

### Code

```python
def minValueNode(self, root):

    current = root

    while current.left is not None:
        current = current.left

    return current
```

---

# 7. Maximum Value

In a BST:

**Maximum = Rightmost Node**

Example:

```text
50
  \
   70
     \
      80
```

Maximum = `80`

### Rule

```text
Go RIGHT → RIGHT → RIGHT → ...
```

### Code

```python
def maxValueNode(self, root):

    current = root

    while current.right is not None:
        current = current.right

    return current
```

# 8. Deletion

Deletion means **removing a node from the BST**.

There are **3 cases**:

1. Node has `0` children
2. Node has `1` child
3. Node has `2` children

---

## Case 1: Node with 0 Children

This is called a **Leaf Node**.

Example:

```text
      50
     /  \
   30    70
```

Delete `30`:

```text
      50
        \
         70
```

Simply remove the node.

---

# Case 2: Node with 1 Child

Example:

```text
      50
     /
   30
     \
      40
```

Delete `30`.

`30` has one child: `40`.

After deletion:

```text
      50
     /
   40
```

### Rule

**Replace the node with its child.**

---

# Case 3: Node with 2 Children

Example:

```text
        50
       /  \
     30    70
          /  \
         60   80
```

Delete `70`.

`70` has two children:

```text
      70
     /  \
   60    80
```

We have two choices:

### Option 1: Inorder Successor

**Minimum value in the RIGHT subtree.**

```text
Go RIGHT → Keep going LEFT
```

For `70`:

```text
Right subtree:

80
```

Therefore:

```text
Successor = 80
```

Replace `70` with `80`, then delete the original `80`.

Result:

```text
        50
       /  \
     30    80
          /
         60
```

---

### Option 2: Inorder Predecessor

**Maximum value in the LEFT subtree.**

```text
Go LEFT → Keep going RIGHT
```

For example:

```text
        70
       /  \
     60    80
       \
        65
```

The maximum value in the left subtree is:

```text
65
```

Therefore:

```text
Predecessor = 65
```

---

# 9. BST Delete Code

Using **Inorder Successor**:

```python
def delete(self, root, key):

    if root is None:
        return None

    # Search left
    if key < root.data:
        root.left = self.delete(root.left, key)

    # Search right
    elif key > root.data:
        root.right = self.delete(root.right, key)

    # Node found
    else:

        # Case 1: No child
        if root.left is None and root.right is None:
            return None

        # Case 2: Only right child
        if root.left is None:
            return root.right

        # Case 2: Only left child
        if root.right is None:
            return root.left

        # Case 3: Two children
        successor = self.minValueNode(root.right)

        root.data = successor.data

        root.right = self.delete(
            root.right,
            successor.data
        )

    return root
```

---

# 10. Understand Two Important Lines

### Line 1

```python
root.right = self.delete(root.right, successor.data)
```

Meaning:

**Go to the right subtree and delete the original successor.**

The function returns the updated right subtree, so we reconnect it to:

```python
root.right
```

---

### Line 2

```python
return root
```

Meaning:

**Return the root of the updated subtree.**

Remember:

```text
Recursive function
       ↓
Update subtree
       ↓
Return updated subtree
       ↓
Parent reconnects it
```

For example:

```python
root.left = self.delete(root.left, key)
```

and:

```python
root.right = self.delete(root.right, key)
```

# 11. Successor vs Predecessor

| Method      | Where to search | What to find |
| ----------- | --------------- | ------------ |
| Successor   | Right subtree   | Minimum      |
| Predecessor | Left subtree    | Maximum      |

Easy memory trick:

```text
SUCCESSOR
    ↓
RIGHT
    ↓
MINIMUM
    ↓
LEFTMOST
```

```text
PREDECESSOR
    ↓
LEFT
    ↓
MAXIMUM
    ↓
RIGHTMOST
```

# 12. BST Operations

| Operation | Purpose             |
| --------- | ------------------- |
| Insert    | Add a node          |
| Search    | Find a value        |
| Delete    | Remove a node       |
| Minimum   | Find smallest value |
| Maximum   | Find largest value  |
| Inorder   | Get sorted values   |

---

# 13. Time Complexity

For a **balanced BST**:

| Operation | Average Time |
| --------- | -----------: |
| Search    |   `O(log n)` |
| Insert    |   `O(log n)` |
| Delete    |   `O(log n)` |
| Minimum   |   `O(log n)` |
| Maximum   |   `O(log n)` |

For a **skewed BST**:

```text
10
  \
   20
     \
      30
        \
         40
           \
            50
```

It behaves like a linked list.

Therefore:

| Operation | Worst Case |
| --------- | ---------: |
| Search    |     `O(n)` |
| Insert    |     `O(n)` |
| Delete    |     `O(n)` |

````python
class Node:
     def __init__(self,value):
        self.left=None
        self.data=value
        self.right=None     
````

````python
from collections import deque        

class BinaryTree:

    def __init__(self):
        self.root = None
    
    def CreateNode(self,data):   # create node
      return Node(data)

    #insertion on node    

    def insert(self,node,value):
        
        if node == None:
            return self.CreateNode(value)
            
        if value < node.data:
             node.left = self.insert(node.left,value)
            
        else:
            node.right = self.insert(node.right,value)

        return node    


    # Traversal in Binary Treee
    
    def traverse_inorder(self,root):  #left,root,right
        if root != None:
            self.traverse_inorder(root.left)
            print(root.data,end=" ")
            self.traverse_inorder(root.right)

    def traverse_preorder(self,root):  #root,left,right
        if root != None:
            print(root.data,end=" ")
            self.traverse_preorder(root.left)
            self.traverse_preorder(root.right)

    def traverse_postorder(self,root):  #left,right,root
        if root != None:  
            self.traverse_postorder(root.left)
            self.traverse_postorder(root.right)   
            print(root.data,end=" ")


    
    def levelwise_traversal(self,root):   # using queue  [BFS]
        if root is None:
            return
        q = []
        q.append(root)
        while len(q) != 0:
            root = q.pop(0)
            print(root.data,end = ' ')

            if root.left is not None:
                q.append(root.left)
            if root.right is not None:
                q.append(root.right)      

    # searching

    def Search(self, node, value):

        if node is None:   #we didn't find the node we were looking for somewhere in the whole tree.
            return False

        if node.data == value:
           return True

        if value < node.data:
           return self.Search(node.left, value)

        return self.Search(node.right, value)


    # deletion    

    def delete(self, root, value):

        # Value not found
        if root is None:
            return None
    
        # Search in left subtree
        if value < root.data:
            root.left = self.delete(root.left, value)
    
        # Search in right subtree
        elif value > root.data:
            root.right = self.delete(root.right, value)
    
        # Node found
        else:
    
            # 1. Leaf node
            if root.left is None and root.right is None:
                return None
    
            # 2. Only right child
            if root.left is None:
                return root.right
    
            # 3. Only left child
            if root.right is None:
                return root.left
    
            # 4. Two children
            successor = root.right
    
            while successor.left is not None:
                successor = successor.left
    
            root.data = successor.data
    
            root.right = self.delete(root.right, successor.data)
    
        return root
````
