```python
class Node:
    def __init__(self,value):
        self.val=value
        self.next=None
        self.prev=None
```

```python

n1=Node(10)
n2=Node(15)
n3=Node(20)
```

```python
n1.next=n2
n2.next=n3
n2.prev=n1
n3.prev=n2
```

```python
class DoublyLL:
    def __init__(self):
        self.head=None
        self.n=0

    #insert at head
    def insert_at_head(self,value):
        new_node = Node(value)
        if not self.head:
            self.head = new_node

            self.n+=1
        else:
            new_node.next= self.head
            self.head.prev= new_node
            self.head = new_node
            self.n+=1

     #add at end
    def append(self,val):
         new_node = Node(val)
         if not self.head:
             self.head = new_node
             self.n+=1
         else :
             curr =self.head
             while curr.next !=None:
                 curr=curr.next
             curr.next=new_node
             new_node.prev=curr
             self.n+=1
             
     #traverse 
    def traverse(self):
        curr = self.head
        while curr != None:
            print(curr.val,end=" ")
            curr=curr.next

    #insert_after
    def insert_after(self,curr_val,item):
        new_node=Node(item)
        curr = self.head
        while curr != None :
            if curr.val==curr_val: 
                break
            curr=curr.next   
        if curr != None:    
           new_node.next=curr.next
           new_node.prev = curr
           self.n+=1
            
           if curr.next !=None:
               curr.next.prev=new_node
               
           curr.next=new_node  
        else:
            return "item not found"

    def size(self):
        curr=self.head
        while curr!=None:
            self.n+=1
        return self.n    
            
            
            
            
         
         
```

```python
dll=DoublyLL()
```

```python
dll.insert_at_head(12)
dll.insert_at_head(14)
dll.insert_at_head(10)
```

```python
dll.append(34)
dll.append(3)
dll.append(337)
```

```python
dll.traverse()
```

```python
dll.insert_after(337,100) 
```
