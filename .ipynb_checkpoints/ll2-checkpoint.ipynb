```python
class Node:
    def __init__(self,val):
        self.data = val
        self.next = None
```

```python
class LinkedList :
    def __init__(self):
        self.head=None
        self.n = 0
        
    def add_head(self,value):
        
        new_node = Node(value)

        new_node.next = self.head
        self.head = new_node
        self.n = self.n + 1

    def __str__(self):

        curr = self.head
        result = ""

        while curr != None :
            result = result + str(curr.data) + ' -> '
            curr = curr.next
        return result[:-3] 

    # find output of the function when head node of following LL is passed as input    

    def fun(self):          #SSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSSS
        if head == None:
            return

        if head != None:
            print(head.data, end=" ")
            fun(head.next)
        print(head.data," ",end=" ") 

    def MaxValueReplace(self,val): #QQQQQQQQQQQQQQQQQQQQQQQQQQQQQQQQQQ
        
        if self.head == None:
            return "Empty LL"
        
        curr = self.head
        max = self.head 

        while curr != None:
           if curr.data> max.data:
               max=curr
           curr = curr.next   

        ppp= max.data
        max.data= val

        return f'Replace max value {ppp} to given value {val}'

        
    def traverse(self):
        curr= self.head
        res=""

        while curr != None:
            res = res + str(curr.data)
            curr = curr.next
            
        return res 


    def OddSum(self):
        curr = self.head
        pos=1
        odd_sum=0
        while curr != None:
            
           
            if pos % 2 != 0:
                odd_sum+=curr.data

            curr=curr.next

            pos+=1

        return odd_sum


    #reverse in place containg integer

    def ReverseLL(self)  :     #3QQQQQQQQQQQQQ

        prev_node = None
        curr_node = self.head

        while curr_node != None:
            next_node = curr_node.next
            curr_node.next = prev_node
            prev_node = curr_node
            curr_node = next_node

        self.head = prev_node


    def change_sent(self):  ##QQQQQQQQQQQQQQ
        temp = self.head

        while temp != None :
  
            if temp.data == "*" or temp.data == "/":
               temp.data = " " 

               if temp.next.data == "*" or temp.next.data == "/":
                    temp.next.next.data = temp.next.next.data.upper()
                    temp.next = temp.next.next

            temp = temp.next
            
                
            
        
       
       
```
