```python
class Node:
    def __init__(self,value) :
        self.data = value 
        self.next = None
    
```

```python
class LinkedList :
    def __init__(self):
        self.head=None
        self.n = 0
        
    def add_head(self,value):
        
        new_node = Node(value)

        new_node.next = self.head
        self.head = new_node
        self.n = self.n + 1

    def __str__(self):

        curr = self.head
        result = ""

        while curr != None :
            result = result + str(curr.data) + ' -> '
            curr = curr.next
        return result[:-3] 

    def search (self,val):

        curr = self.head
        i=0

        while curr != None:
            if curr.data == val:
               return f'index = {i}'
               
            curr = curr.next
            i+=1
        return 'Not found'    

    def search_ind(self,ind):

        curr = self.head
        i=0

        while curr != None:
            if ind == i :
                return curr.data
            curr = curr.next
            i+=1
        return 'Not found'    
            

            


     
```

```python
l=LinkedList()
```

```python
l.add_head(23)
l.add_head(33)
l.add_head(453)
l.add_head(3)
l.add_head(30)
```

```python
l.search(30)
```
